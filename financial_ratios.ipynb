```python
def calculate_per(price, eps):
  return price / eps

def calculate_dividend_yield(dividend, price):
  return dividend / price * 100

def calculate_roe(net_income, equity):
  return net_income / equity * 100

price = 100
eps = 5
dividend = 3
net_income = 2000
equity = 10000

print("PER:", calculate_per(price, eps))

print("Dividend Yield:", calculate_dividend_yield(dividend, price))

print("ROE:", calculate_roe(net_income, equity))


```
